```python
import pandas as pd
import numpy as np
```

```python
# Load your existing CSV
df = pd.read_csv('final_data.csv')
```

```python

# Add new features with random plausible values
np.random.seed(42)  # for reproducibility
```

```python
# Sleep hours: 4 to 10
df['Sleep_Hours'] = np.random.randint(4, 11, size=len(df))
```

```python
# Water intake: 1 to 4 liters
df['Water_Intake'] = np.round(np.random.uniform(1, 4, size=len(df)), 1)
```

```python
# Diet preference
df['Diet_Preference'] = np.random.choice(['Vegetarian', 'Non-Vegetarian', 'Vegan'], size=len(df))
```

```python
# Stress level
df['Stress_Level'] = np.random.choice(['Low', 'Moderate', 'High'], size=len(df))
```

```python
# Health condition
df['Health_Condition'] = np.random.choice(['None', 'Diabetes', 'BP', 'Cholesterol', 'Other'], size=len(df))
```

```python
# Save the updated CSV
df.to_csv('final_data_updated.csv', index=False)
```

```python
print("Updated CSV created: final_data_updated.csv")
```
